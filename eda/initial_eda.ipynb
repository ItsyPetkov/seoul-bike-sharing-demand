Check the types of the selected columns for any inconsistencies

```r
for(dataset_name in dataset_list){
    dataset <- read_csv(dataset_name)
    print(
        dataset %>% summarise(across(everything(), class)) %>%
      pivot_longer(cols = everything(), names_to = "variable", values_to = "class")
    )
}
```

They are all interpreted as character columns, but I expect the `BICYCLES` and `DAILY_RIDERSHIP` column to be of numeric type. Let's see why it wasn't loaded as a numeric column - possibly some entries contain characters.

```r
# grepl searches a string for non-digital characters, and returns TRUE or FALSE
# if it finds any non-digital characters, then a column is not purely numeric
find_character <- function(strings) grepl("[^0-9]", strings)
```

```r
dataset <- read_csv('../data_raw/raw_bike_sharing_systems.csv')
dataset %>% 
    select(BICYCLES) %>% 
    filter(find_character(BICYCLES)) %>%
    slice(0:10)
```

As can be seen, many rows have non-numeric characters, such as `32 (including 6 rollers) [162]` and `1000[253]`. This is actually very common for a table scraped from Wiki when no input validation is enforced.

Next, I take a look at the other columns, namely `COUNTRY`, `CITY`, and `SYSTEM`, to see if they contain any undesired reference links, such as `Melbourne[12]` for example.

```r
# Defined a 'reference link' character class, 
# `[A-z0-9]` means at least one character 
# `\\[` and `\\]` means the character is wrapped by [], such as for [12] or [abc]
ref_pattern <- "\\[[A-z0-9]+\\]"
find_reference_pattern <- function(strings) grepl(ref_pattern, strings)
```

```r
# Check whether the COUNTRY column has any reference links
dataset %>% 
    select(COUNTRY) %>% 
    filter(find_reference_pattern(COUNTRY)) %>%
    slice(0:10)
```

Ok, looks like the `COUNTRY` column is clean. Let's check the `CITY` column.

```r
# Check whether the CITY column has any reference links
dataset %>% 
    select(City) %>% 
    filter(find_reference_pattern(City)) %>%
    slice(0:10)
```

Looks like the `CITY` column has some reference links to be removed. Next, let's check the `SYSTEM` column.

```r
# Check whether the System column has any reference links
dataset %>% 
    select(SYSTEM) %>% 
    filter(find_reference_pattern(SYSTEM)) %>%
    slice(0:10)
```

So the `SYSTEM` column also has some reference links.

After some preliminary investigations, we identified that the `CITY` and `SYSTEM` columns have some undesired reference links, and the `BICYCLES` column has both reference links and some textual annotations.

Additionally, there are a lot of missing values throughout all datasets, so they need to be addressed.

The contents of the `HOUR` column are stored as numeric, where infact it should be character as the hours represent categories.
